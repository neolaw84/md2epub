## Step 1: Brainstorming Inputs
The story inputs are hard-coded for this comprehensive demonstration.

```python
premise = """In the smog-choked heart of Victorian London, a disgraced astronomer named Arthur Penhaligon discovers a terrifying truth that defies every known law of physics: the stars are not distant, burning spheres of gas but are, in fact, the colossal, lidless eyes of incomprehensible biological entities peering through the thin veil of our reality. As he tracks their rhythmic blinking from a derelict mountaintop observatory, he realizes that the celestial bodies are slowly focusing their collective gaze upon the Earth, preparing to 'blink' in unison—an event that would effectively shatter the foundations of our physical existence and replace them with the chaotic whims of these cosmic watchers. Arthur must race against his own dwindling sanity and a shadowy cult to find a way to 'blind' the sky, even if it means plunging the world into eternal darkness and ensuring that humanity never looks at the heavens with wonder again."""

genre = """This narrative is a meticulously crafted fusion of Victorian Gothic Horror, Cosmic Mystery, and early Steampunk aesthetics. It draws heavy inspiration from the existential dread of H.P. Lovecraft while maintaining the grounded, character-driven drama of Mary Shelley's Frankenstein. The story leans into the 'Weird Fiction' subgenre, emphasizing atmosphere, suspense, and the crushing weight of the unknown. The tone should transition from a scholarly, scientific curiosity to a frantic, fever-dream-like desperation as the scale of the threat becomes apparent. It is a world where the hiss of steam and the click of gears are the only defense against the silent, suffocating presence of entities that existed before time and will persist long after humanity has been forgotten by the stars. The pacing must be relentless, mirroring the ticking clock of a universe about to be reset."""

setting = """The story unfolds across two primary locations: the labyrinthine, fog-shrouded streets of 1888 London and the desolate, wind-swept ridges of the Scottish Highlands. London is depicted as a city of harsh contrasts—where the opulence of the Royal Society clashes with the squalor of the East End, all unified by the pervasive grime of the industrial revolution. In contrast, the Highlands location is the 'Blackwood Peak Observatory,' an ancient stone structure retrofitted with experimental, brass-bound clockwork mechanisms that groan under the strain of their forbidden purpose. The environment is a character in itself, with gaslights that flicker in the presence of celestial attention and skies that seem to pulse with a sickly, violet hue that only those who have 'seen' can truly perceive, creating a pervasive sense of being watched from above."""

themes = """At its core, the narrative explores the devastating burden of forbidden knowledge and the absolute collapse of human reason when faced with the vastness of an indifferent universe. It examines the theme of 'Solipsism vs. Reality,' questioning whether knowing the truth is worth the price of one's mind. The corruption of power is also central, as represented by the aristocratic cultists who view the impending cosmic apocalypse as a deserved 'cleansing' of the lower classes, reflecting 19th-century fears of social unrest and the dehumanizing nature of extreme hierarchy. Finally, it deals with the concept of 'The Sacrifice of Information,' where the only way to save humanity is to ensure that the secrets of the universe remain buried forever, even if it means Arthur is remembered only as a madman who tried to destroy the science he once loved."""

key_protagonists = """Arthur Penhaligon is the protagonist, a man defined by a toxic mixture of intellectual arrogance and a desperate need for vindication after his peers branded him a lunatic. He is physically frail but possesses a sharp, obsessive mind. His foil and antagonist is Lord Alistair Vane, a charismatic but sociopathic nobleman who leads 'The Ocular Order.' Vane believes the stars are gods to be worshipped, and he is willing to commit any atrocity to ensure their gaze is never diverted. Supporting them is Elspeth Thorne, a brilliant, soot-stained clockwork engineer who lost her father to the same 'star-sickness' and provides the mechanical expertise Arthur lacks. Adding to the tension is 'The Silent Warden,' a masked enforcer of the Order who never speaks but is always present in the periphery of Arthur’s vision, representing the physical threat of the cult and the silent monitoring of the celestial eyes."""

additional_requests = """I request that the narrative style adopts the formal, slightly archaic, and highly descriptive prose characteristic of late Victorian literature, utilizing complex sentence structures and a rich vocabulary to evoke the era's atmosphere. Special attention should be paid to the sensory details of the observatory: the smell of ozone and old leather, the deafening roar of the Scottish wind, and the terrifying, silent intensity of the 'Gaze.' The pacing should be a deliberate slow-burn, building a sense of inescapable doom until the final act. I specifically want the 'Logistics Manager' agent to track the physical condition of the 'Ocular Lens' (which is cracking under the strain) and Arthur’s supply of laudanum, which he uses to suppress the visions but which also clouds his judgment at critical moments. The ending should be bittersweet, acknowledging that while the world is saved, the cost has been Arthur’s soul and legacy. I want to emphasize the Lovecraftian scale of the entities."""

output_path = "cosmic_horror_story"

# Combine extra info
additional_info_block = f"""
Setting: {setting}
Themes: {themes}
Key Characters: {key_protagonists}
User Notes: {additional_requests}
"""
```

## Step 2: Initialize Agents
We are setting up the LangGraph workflow.

```python
app = create_story_graph()
```

## Step 3: Run the Agents with Configuration
Following SOLID principles, we pass Infrastructural settings (LLM params) via the graph config, keeping the Domain State (Story inputs) clean.

```python
# Technical Configuration (Infrastructural)
config = {
    "configurable": {
        "llm_params": {
            "model_name": "gpt-4o",
            "temperature": 0.8
        },
        "max_iterations": 2
    }
}

# Domain Inputs (Story Content)
inputs = {
    "premise": premise,
    "genre": genre,
    "additional_info": additional_info_block,
    "user_output_dir": output_path,
    "iteration": 0
}

print("Starting generation process...")

for output in app.stream(inputs, config=config):
    for key, value in output.items():
        if key == "EditorCritique":
            critique = value.get("critique")
            if critique:
                print(f"\n>>> EDITOR CRITIQUE: Iterating...\n{critique[:200]}...\n")
            else:
                print("\n>>> EDITOR APPROVED. Finalizing...")
        print(f"Finished step: {key}")

print(f"\nDone! All files generated in {output_path}/story-elements/")
```
